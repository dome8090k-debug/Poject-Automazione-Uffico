## 1. Creare il primo file Excel

In `openpyxl`:

- `Workbook()` crea un **nuovo file Excel**.
- Il file contiene uno o più **fogli di lavoro** (*worksheet*).
- Il foglio attivo è quello creato automaticamente all'apertura del workbook.
- Ogni cella può essere indicata con la sua coordinata, ad esempio `A1`, `B2`, `C3`.

```python
# Importiamo la classe Workbook dalla libreria openpyxl
# Workbook rappresenta un intero file Excel
from openpyxl import Workbook

# Creiamo un nuovo file Excel vuoto in memoria
wb = Workbook()

# Selezioniamo il foglio attivo
# Quando creiamo un nuovo Workbook, openpyxl crea automaticamente un foglio
ws = wb.active

# Cambiamo il nome del foglio attivo
# Il nome apparirà nella scheda in basso in Excel
ws.title = "Vendite"

# Scriviamo l'intestazione della colonna A, cella A1
ws["A1"] = "Prodotto"

# Scriviamo l'intestazione della colonna B, cella B1
ws["B1"] = "Quantità"

# Scriviamo l'intestazione della colonna C, cella C1
ws["C1"] = "Prezzo unitario"

# Scriviamo l'intestazione della colonna D, cella D1
ws["D1"] = "Totale"

# Salviamo il file Excel nella cartella output
# Se il file esiste già, viene sovrascritto
wb.save("output/primo_file.xlsx")

# Stampiamo un messaggio di conferma
print("File creato: output/primo_file.xlsx")
```

### Inserire più righe con `append()`

Invece di scrivere ogni cella separatamente, possiamo usare `append()`.

`append()` aggiunge una **riga completa** al foglio Excel.

La prima lista sarà l'intestazione; le altre liste saranno i dati.

```python
# Importiamo di nuovo Workbook
# In un notebook possiamo reimportare le librerie senza problemi
from openpyxl import Workbook

# Creiamo un nuovo file Excel
wb = Workbook()

# Selezioniamo il foglio attivo
ws = wb.active

# Assegniamo un nome al foglio
ws.title = "Vendite"

# Creiamo una lista di liste
# Ogni lista interna rappresenta una riga di Excel
dati = [
    # Prima riga: intestazioni delle colonne
    ["Prodotto", "Quantità", "Prezzo unitario", "Totale"],

    # Seconda riga: primo prodotto
    ["Tour in barca", 2, 120, None],

    # Terza riga: secondo prodotto
    ["Transfer aeroporto", 1, 45, None],

    # Quarta riga: terzo prodotto
    ["Guida turistica", 3, 30, None],

    # Quinta riga: quarto prodotto
    ["Degustazione", 4, 25, None],
]

# Cicliamo su ogni riga della lista 'dati'
for riga in dati:
    # Aggiungiamo la riga al foglio Excel
    # La prima chiamata a append() scrive nella riga 1
    # La seconda nella riga 2, e così via
    ws.append(riga)

# Salviamo il file Excel
wb.save("output/vendite.xlsx")

# Confermiamo il salvataggio
print("File salvato: output/vendite.xlsx")
```

**Nota importante:**  

`None` lascia la cella vuota.  

Nella colonna `Totale` non abbiamo scritto un numero perché il totale verrà calcolato successivamente con una formula Excel.

## 2. Leggere un file Excel esistente

Per aprire un file già esistente si usa `load_workbook()`.

- `Workbook()` crea un file nuovo.
- `load_workbook()` apre un file esistente.

Una volta aperto il file, possiamo:

- Vedere quali fogli contiene.
- Leggere il valore di una singola cella.
- Leggere tutte le righe con `iter_rows()`.

```python
# Importiamo la funzione load_workbook
# Serve per aprire un file Excel già esistente
from openpyxl import load_workbook

# Apriamo il file creato in precedenza
wb = load_workbook("output/vendite.xlsx")

# Selezioniamo il foglio chiamato 'Vendite'
ws = wb["Vendite"]

# Stampiamo l'elenco di tutti i fogli presenti nel file
print("Fogli presenti:", wb.sheetnames)

# Leggiamo e stampiamo il valore della cella A1
print("Valore cella A1:", ws["A1"].value)

# Leggiamo e stampiamo il valore della cella B2
print("Valore cella B2:", ws["B2"].value)
```

```python
# Apriamo nuovamente il file Excel
wb = load_workbook("output/vendite.xlsx")

# Selezioniamo il foglio 'Vendite'
ws = wb["Vendite"]

# iter_rows() permette di scorrere le righe del foglio
# min_row=1 indica di partire dalla prima riga
# values_only=True restituisce direttamente i valori delle celle
# Senza values_only=True, otterremmo oggetti Cell e non i valori
for riga in ws.iter_rows(min_row=1, values_only=True):
    # Stampiamo la riga sotto forma di tupla
    print(riga)
```

### Leggere solo i dati

Spesso la prima riga contiene le intestazioni e non ci interessa leggerla come dato.

Per iniziare dalla seconda riga usiamo `min_row=2`.

```python
# Apriamo il file Excel
wb = load_workbook("output/vendite.xlsx")

# Selezioniamo il foglio
ws = wb["Vendite"]

# Scorriamo le righe a partire dalla riga 2
# Così evitiamo di leggere l'intestazione
for prodotto, quantita, prezzo, totale in ws.iter_rows(min_row=2, values_only=True):
    # Stampiamo una frase formattata con i dati della riga
    print(f"{prodotto}: {quantita} x {prezzo} €")
```

## 3. Aggiungere formule Excel

Excel può calcolare automaticamente i valori tramite formule.

In Python, una formula Excel è semplicemente una stringa che inizia con `=`.

Esempi:

- `=B2*C2` moltiplica la quantità per il prezzo.
- `=SUM(D2:D5)` somma i valori della colonna D dalla riga 2 alla riga 5.

```python
# Apriamo il file Excel esistente
wb = load_workbook("output/vendite.xlsx")

# Selezioniamo il foglio
ws = wb["Vendite"]

# max_row restituisce il numero dell'ultima riga che contiene dati
ultima_riga = ws.max_row

# Cicliamo dalle righe 2 fino all'ultima riga inclusa
# range(2, ultima_riga + 1) genera: 2, 3, 4, 5
for riga in range(2, ultima_riga + 1):
    # Selezioniamo la cella della colonna 4, cioè la colonna D
    # column=1 -> A, column=2 -> B, column=3 -> C, column=4 -> D
    cella_totale = ws.cell(row=riga, column=4)

    # Scriviamo una formula Excel
    # La formula moltiplica la quantità (colonna B) per il prezzo (colonna C)
    cella_totale.value = f"=B{riga}*C{riga}"

# Calcoliamo la riga successiva all'ultima riga di dati
riga_totale = ultima_riga + 1

# Scriviamo l'etichetta 'TOTALE' nella colonna C
ws.cell(row=riga_totale, column=3).value = "TOTALE"

# Scriviamo la formula che somma tutti i totali della colonna D
ws.cell(row=riga_totale, column=4).value = f"=SUM(D2:D{ultima_riga})"

# Salviamo il file con un nuovo nome
wb.save("output/vendite_con_totali.xlsx")

# Confermiamo l'operazione
print("Formule aggiunte e file salvato.")
```

```python
# Apriamo il file Excel esistente
wb = load_workbook("output/vendite.xlsx")

# Selezioniamo il foglio
ws = wb["Vendite"]

# Aggiungiamo l'intestazione della nuova colonna E
ws["E1"] = "Totale scontato 10%"

# Cicliamo dalle righe 2 fino all'ultima riga inclusa
# range(2, ultima_riga + 1) genera: 2, 3, 4, 5
for riga in range(2, ultima_riga + 1):
  # Selezioniamo la cella della colonna E, cioè il totale scontato
    # column=5 corrisponde alla colonna E
    cella_scontata = ws.cell(row=riga, column=5)
    # Scriviamo la formula del totale scontato
    # Il totale normale viene moltiplicato per 0.90
    # perché uno sconto del 10% equivale a pagare il 90% del prezzo
    cella_scontata.value = f"=D{riga}*0.9"

# Calcoliamo la riga successiva all'ultima riga di dati
riga_totale = ultima_riga + 1
# Scriviamo la formula che somma tutti i totali scontati della colonna E
ws.cell(row=riga_totale, column=5).value = f"=SUM(E2:E{ultima_riga})"

# Applichiamo il formato valuta alla colonna E
for riga in range(2, riga_totale + 1):
    ws.cell(row=riga, column=5).number_format = '#,##0.00 €'

# Salviamo il file con un nuovo nome
wb.save("output/vendite_con_sconto.xlsx")
```

**Attenzione:**  

`openpyxl` scrive la formula, ma non calcola il risultato.

Il risultato verrà calcolato quando apriremo il file con Excel, LibreOffice o Google Sheets.

## 4. Formattare il file Excel

Con `openpyxl` possiamo migliorare l'aspetto del report:

- Grassetto e colore del testo con `Font`.
- Colore di sfondo con `PatternFill`.
- Allineamento con `Alignment`.
- Larghezza delle colonne con `column_dimensions`.
- Formato numerico, ad esempio valuta, con `number_format`.

```python
# Importiamo load_workbook per aprire il file
from openpyxl import load_workbook

# Importiamo gli stili necessari
from openpyxl.styles import Font, PatternFill, Alignment

# Apriamo il file con le formule
wb = load_workbook("output/vendite_con_totali.xlsx")

# Selezioniamo il foglio
ws = wb["Vendite"]

# Creiamo uno stile per il font dell'intestazione
# bold=True rende il testo in grassetto
# color="FFFFFF" imposta il colore bianco
# size=12 imposta la dimensione del carattere
intestazione_font = Font(bold=True, color="FFFFFF", size=12)

# Creiamo un riempimento blu per lo sfondo delle celle
# fill_type="solid" indica un colore pieno
# start_color è il colore esadecimale scelto
intestazione_fill = PatternFill(fill_type="solid", start_color="2563EB")

# Cicliamo sulle prime 4 colonne: A, B, C e D
# range(1, 5) genera 1, 2, 3, 4
for colonna in range(1, 5):
    # Selezioniamo la cella dell'intestazione nella riga 1
    cella = ws.cell(row=1, column=colonna)

    # Applichiamo il font creato in precedenza
    cella.font = intestazione_font

    # Applichiamo il colore di sfondo
    cella.fill = intestazione_fill

    # Centriamo il testo nella cella
    cella.alignment = Alignment(horizontal="center")

# Impostiamo la larghezza della colonna A
ws.column_dimensions["A"].width = 25

# Impostiamo la larghezza della colonna B
ws.column_dimensions["B"].width = 15

# Impostiamo la larghezza della colonna C
ws.column_dimensions["C"].width = 20

# Impostiamo la larghezza della colonna D
ws.column_dimensions["D"].width = 20

# Formattiamo le colonne C e D come valuta
# Cicliamo dalla riga 2 fino all'ultima riga con dati
for riga in range(2, ws.max_row):
    # Applichiamo il formato euro alla colonna C
    ws.cell(row=riga, column=3).number_format = '#,##0.00 €'

    # Applichiamo il formato euro alla colonna D
    ws.cell(row=riga, column=4).number_format = '#,##0.00 €'

# Salviamo il file formattato
wb.save("output/vendite_formattate.xlsx")

# Confermiamo il salvataggio
print("Formattazione completata.")
```

```python
excel_vendite_sconto = load_workbook("output/vendite_con_sconto.xlsx")
excel_vendite_sconto_foglio = excel_vendite_sconto["Vendite"]

intestazione_sconto_font = Font(bold=True, color='000000', size=18)
intestazione_sconto_fill = PatternFill(fill_type='solid', start_color='FFE4E1')

for colonna in range(1,6):
  cella=excel_vendite_sconto_foglio.cell(row=1, column=colonna)
  cella.font = intestazione_sconto_font
  cella.fill = intestazione_sconto_fill
  cella.alignment = Alignment(horizontal='center')
# Salviamo il file formattato
wb.save("output/vendite_con_sconto_formattate.xlsx")

# Confermiamo il salvataggio
print("Formattazione completata.")
```

### Evidenziare automaticamente alcune celle

Possiamo cambiare colore o font in base al valore contenuto in una cella.

In questo esempio evidenziamo in giallo i totali superiori a 200 €.

**Nota:** le formule Excel non hanno un valore numerico finché il file non viene aperto e calcolato da Excel. Per questo motivo, in questo esempio controlliamo solo i valori già presenti come numeri.

```python
# Apriamo il file Excel formattato
wb = load_workbook("output/vendite_formattate.xlsx")

# Selezioniamo il foglio
ws = wb["Vendite"]

# Creiamo un riempimento giallo
giallo = PatternFill(fill_type="solid", start_color="FFF200")

# Creiamo un font rosso e in grassetto
rosso = Font(color="CC0000", bold=True)

# Scorriamo le righe dei dati
for riga in range(2, ws.max_row):
    # Leggiamo il valore della colonna D, cioè il totale
    totale = ws.cell(row=riga, column=4).value

    quantita = ws.cell(row=riga, column=2).value
    prezzo = ws.cell(row=riga, column=3).value
    totale_calcolato=quantita*prezzo

    # isinstance() controlla se 'totale' è un numero intero o decimale
    # Questo evita errori se la cella contiene una formula
    if totale_calcolato > 200:
        # Applichiamo lo sfondo giallo alla cella del totale
        ws.cell(row=riga, column=4).fill = giallo

        # Applichiamo il font rosso alla stessa cella
        ws.cell(row=riga, column=4).font = rosso

# Salviamo il file finale
wb.save("output/vendite_evidenziate.xlsx")

# Confermiamo l'operazione
print("Evidenziazione completata.")
```
